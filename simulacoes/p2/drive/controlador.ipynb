```python
import math as mt
P = 4
l1 = 0.0023
l2 = 0.0036
lm = 0.0607
r1 = 0.1975
r2 = 0.1909
vs = 400/mt.sqrt(3)
i2 = 24.4388
f = 50
B = 0.0103
J = 0.095

i = 10
ksv = (2/100) * (60/(mt.pi*2))
Jload = 24.5
Bload = 4.04

Jeq = J + (Jload/(i**2))
Beq = B + (Bload/(i**2))
print (Beq)
print (f"J = {Jeq}")
```

```python
kvf = (vs - (i2*r1)) / (2*mt.pi*f*(1 + (l1/lm)))
# kvf = vs/(2*mt.pi*f)
ksl = 3*(kvf**2)*((P/2)**2)*(1/r2)

bloco = kvf*(1 + (l1/lm))

print (f"kvf = {kvf}")
print (f"ksl = {ksl}")
print (f"Bloco = {bloco}")
print (f"I2*R1 = {i2*r1}")
```

```python
Kma = (ksl/Beq)*ksv*(1/i)
tau_ma = Jeq/Beq

tau_mf = 1.2/3
print (f"tma = {tau_ma}")
print (f"kma = {Kma}")
```

```python
pss = 5
zeta = mt.sqrt( (mt.log(pss/100.0)**2) / (mt.pi**2 + (mt.log(pss/100.0)**2)) )
zeta
```

```python
Kp = (1/Kma)*((2*tau_ma/tau_mf) - 1)
Ki = (tau_ma/Kma)*(1/((zeta**2)*(tau_mf**2)))

a = Ki/Kp
print (f"Kp = {Kp}")
print (f"Ti = {1/Ki}")
print (f"Tz = {1/a}")
```
